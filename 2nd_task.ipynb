```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import statistics

df=pd.read_csv("cleaned_iris_dataset.csv")
display(df)


numerics=df.select_dtypes("number")
stats = pd.DataFrame({
    "mean": numerics.mean(),
    "median": numerics.median(),
    "mode": numerics.mode().iloc[0],
    "std": numerics.std()
})
print(stats)


numerics.hist(bins=20)
plt.show()

fig, axes = plt.subplots(1, len(numerics.columns), figsize=(14, 4))

for ax, col in zip(axes, numerics.columns):
    sns.boxplot(data=df, x="species", y=col, ax=ax)
    ax.set_title(col)

plt.tight_layout()
plt.show()
```
