task 1 : cleaning data and preprocessing

```python
import pandas as pd
#loading the dataset
df = pd.read_csv("sentiment dataset.csv")
display(df.head())

print(df.isna().sum())
print(df.duplicated().sum())

#deleting unnamed columns 
df= df.drop(columns=[col for col in df.columns if col.startswith('Unnamed') or col== ""])


#removing duplicates
df = df.drop_duplicates()

#removing null values
df = df.dropna()

#format of time column
df['Timestamp'] = pd.to_datetime(df['Timestamp'], errors='coerce')

#verifying the changes
print(df.info())

#saving the cleaned dataset
df.to_csv("cleaned_sentiment_dataset.csv", index=False)
```

```python
df = pd.read_csv("iris.csv")
display(df.head())

print(df.isna().sum())
print(df.duplicated().sum())

#removing duplicates
df=df.drop_duplicates()

#removing missing values
df=df.dropna()

print(df.info())

#saving the cleaned dataset
df.to_csv("cleaned_iris_dataset.csv", index=False)
```

```python
df= pd.read_csv("Stock Prices Data Set.csv")
display(df.head())

#removing duplicates
df=df.drop_duplicates()

#removing missing values
df=df.dropna()

#verifying format
df["date"]=pd.to_datetime(df["date"], errors='coerce')

print(df.info())

#saving
df.to_csv("cleaned_stock_prices_dataset.csv", index=False)
```

```python
#naming the column
cols = ["CRIM","ZN","INDUS","CHAS","NOX","RM","AGE","DIS","RAD","TAX","PTRATIO","B","LSTAT","MEDV"]#Boston Housing dataset, one of the most widely used datasets in statistics and machine learning
df=pd.read_csv("house Prediction Data Set.csv",sep=r"\s+",names=cols,header=None)
display(df.head())

print(df.isna().sum())
print(df.duplicated().sum())

#removing duplicates
df=df.drop_duplicates()

#removing missing values
df=df.dropna()

#saving 
df.to_csv("cleaned_house_prediction.csv")
```
