task3:basic data visualisation

```python
import pandas as pd

import seaborn as sns

df=pd.read_csv("cleaned_sentiment_dataset.csv")
display(df)
```

```python
import matplotlib.pyplot as plt

#barplot (sentiment by platform)
plt.figure(figsize=(8, 15))
sns.countplot(data=df,y="Platform",hue="Sentiment")
plt.title("number of platform by sentiment")
plt.xlabel("platform")
plt.ylabel("number")
plt.legend(title="Sentiment")
plt.savefig("images/barplot_sentiment.png", dpi=150)
plt.show()
```

```python
#linechart (retweets by month)
sns.lineplot(data=df,x="Retweets",y="Month")
plt.title("Number of posts per month")
plt.xlabel("Number of retweets")
plt.ylabel("month")
plt.legend()
plt.savefig("images/linechart_sentiment.png",dpi=150)
plt.show()
```

we can see that numbers of posts is bigger in the month 10 then the others

```python
#scatter plot( likes vs retweets)
sns.scatterplot(data=df,x="Likes",y="Retweets")
plt.title("Likes vs Retweets")
plt.xlabel("number of likes")
plt.ylabel("number of posts")
plt.savefig("images/scatterplot_sentiment.png",dpi=150)
plt.show()
```

we notice that when the number of post gets bigger , the number of likes gets bigger too
