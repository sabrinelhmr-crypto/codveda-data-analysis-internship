task1 : Regression Analysis

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import r2_score, mean_squared_error


df=pd.read_csv("cleaned_house_prediction.csv")
display(df)


```

```python
#spliting dataset into training and testing sets

x=df[["RM"]]
y=df[["MEDV"]]

x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2,random_state=42)
```

```python
#linear regression model

x1=LinearRegression().fit(x_train,y_train)
predict = x1.predict(x_test)

print(f"coefficient: {x1.coef_[0]} and Intercept: {x1.intercept_} ")
print(f"R2 : {r2_score(y_test,predict)}  mean squared error: {mean_squared_error(y_test,predict)} ")
```

```python
#presenting the linear regression on a scatter plot

plt.scatter(x_test,y_test)
plt.plot(x_test,predict,color="red", label="regression line")
plt.title("avg rooms vs price")
plt.xlabel("RM")
plt.ylabel("MEDV")
plt.show
```
